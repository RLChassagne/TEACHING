```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata
# Données
points = np.array([
    [0, 0, 7.1],
    [50, 10, 6.8],
    [100, 0, 7.6],
    [150, 20, 8.0],
    [200, 0, 9.2],
    [25, 50, 6.3],
    [125, 60, 7.9],
    [75, 90, 8.5],
    [175, 80, 9.0],
    [225, 60, 10.1],
    [250, 20, 10.4],
    [300, 50, 11.0]
])

x = points[:, 0]
y = points[:, 1]
z = points[:, 2]

# Grille régulière
xi = np.linspace(min(x), max(x), 200)
yi = np.linspace(min(y), max(y), 200)
X, Y = np.meshgrid(xi, yi)

# Interpolation (cubic / linear / nearest)
Z = griddata((x, y), z, (X, Y), method='linear')

# Affichage
plt.figure(figsize=(8, 6))
contour = plt.contourf(X, Y, Z, levels=20, cmap="viridis")
plt.scatter(x, y, c=z, edgecolor='k', cmap="viridis", s=80, marker="o")
plt.colorbar(contour, label="Porosity (%)")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Porosity Interpolation Map")
plt.show()
```
