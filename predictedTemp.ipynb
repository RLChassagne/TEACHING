```python
import matplotlib.pyplot as plt
import numpy as np

g = [[2, 1], [8, 1]]
g_inv=np.linalg.inv(g)
print('G_inv=', g_inv)

d= np.array([[19], [22]])

#product G_inv with d
m=np.dot(g_inv, d)

print('m=', m)

# Related to plotting
x = np.linspace(0,18,10)
y = m[0]*x + m[1]
fig = plt.figure()
plt.plot(x,y, 'g', label ="Model")
plt.xlabel('Depth (m)')
plt.ylabel('Temperature degree celsius')
plt.plot(2, 19, 'o', color ='blue', label ="data" )
plt.plot(8, 22, 'o', color ='blue', )
plt.plot(15, 25.5, 'o', color ='orange', label ="")
plt.legend() 
plt.show()
```
